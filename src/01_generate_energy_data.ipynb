# Generate Energy Data

```python
from utils.generate_energy_data import generate_timeseries, create_energy_data_csv
from datetime import datetime

start = "2025-01-01"
end = "2026-01-01"
```

### Energy Community

```python
energy_community_consumption = generate_timeseries("energy_community_consumption", "consumption", start, end, seed=42, scale_factor=50)
energy_community_generation = generate_timeseries("energy_community_generation", "generation", start, end, seed=42, scale_factor=50)

energy_data = energy_community_consumption + energy_community_generation

file_path = create_energy_data_csv("energy_community", energy_data)

print(f"Created: {file_path}")
```
