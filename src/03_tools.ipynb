```python
from typing import Literal
import pandas as pd

TimeseriesOperation = Literal[
    "add",
    "subtract",
    "multiply",
    "divide",
]

def calculate_timeseries(
    operation: TimeseriesOperation,
    timeseries: list[pd.DataFrame],
    scalar: float | None = None,
) -> pd.DataFrame:
    """
    Perform an arithmetic operation on energy time series.

    For addition and subtraction, multiple time series can be combined.
    For multiplication and division, a time series can be combined
    with a scalar.
    """

    if not timeseries:
        raise ValueError("At least one time series is required.")

    if operation in ("add", "subtract"):

        if len(timeseries) < 2:
            raise ValueError(
                f"Operation '{operation}' requires at least two time series."
            )

        result = timeseries[0].copy()

        for other in timeseries[1:]:
            if operation == "add":
                result["value_kwh"] = (
                    result["value_kwh"] + other["value_kwh"]
                )

            elif operation == "subtract":
                result["value_kwh"] = (
                    result["value_kwh"] - other["value_kwh"]
                )

        return result

    if operation in ("multiply", "divide"):

        if len(timeseries) != 1:
            raise ValueError(
                f"Operation '{operation}' requires exactly one time series."
            )

        if scalar is None:
            raise ValueError(
                f"Operation '{operation}' requires a scalar."
            )

        result = timeseries[0].copy()

        if operation == "multiply":
            result["value_kwh"] = (
                result["value_kwh"] * scalar
            )

        elif operation == "divide":
            if scalar == 0:
                raise ValueError("Division by zero is not allowed.")

            result["value_kwh"] = (
                result["value_kwh"] / scalar
            )

        return result

    raise ValueError(
        f"Unknown operation: {operation}"
    )
```
