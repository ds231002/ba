# Generate Energy Data

```python
from utils.generate_energy_data import generate_timeseries, create_energy_data_csv
```

### Timerange

```python
start = "2025-01-01"
end = "2026-01-01"
```

### Energy Community

```python
community_consumption = generate_timeseries(
    "community_consumption",
    "consumption", start,
    end, seed=42,
    scale_factor=50
)

community_generation = generate_timeseries(
    "community_generation",
    "generation",
    start,
    end,
    seed=42,
    scale_factor=50
)

energy_data = community_consumption + community_generation

file_path = create_energy_data_csv("community", energy_data)

print(f"Created: {file_path}")
```

### User

```python
mp_1 = generate_timeseries(
    "mp_1", "consumption",
    start,
    end,
    seed=42
)

mp_2 = generate_timeseries(
    "mp_2",
    "consumption",
    start,
    end,
    seed=43
)

mp_3 = generate_timeseries(
    "mp_3",
    "generation",
    start,
    end,
    seed=44
)

energy_data = mp_1 + mp_2 + mp_3

file_path = create_energy_data_csv("user", energy_data)

print(f"Created: {file_path}")
```
